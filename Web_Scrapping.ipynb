#Sqlite

```python
import sqlite3
conn = sqlite3.connect("students.db")
```

```python
cursor =  conn.cursor()
print("DataBase Created Successfully")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS student_table(
NAME TEXT,
ROLL_NO INTEGER,
MARKS INTEGER )
"""
)
print("Table is created successfully !!")
```

```python
#write the data into table :-
cursor.execute(
    "INSERT INTO student_table VALUES (? ,? ,?)",
    ("Amit",101,90)
)
cursor.execute(
    "INSERT INTO student_table VALUES ('Aman',102,95) "
)
```

```python
conn.commit()
print("Records inserted successfully !!")
conn.close()
```

```python
#Retrive | read the data
conn = sqlite3.connect("students.db")
cursor = conn.cursor()
cursor.execute("SELECT * FROM student_table")
rows = cursor.fetchall()
for row in rows :
  print(row)
  conn.close()
```
